# Random tensors

## Why random tensors?

### Random tensors are important in deep learning because they help us initialize neural network weights, test models, generate synthetic data, and introduce   randomness during training.

```python
# Create a random tensor of size(3, 4)

random_tensor = torch.rand(1, 3, 4)
random_tensor
```

```python
random_tensor.ndim, random_tensor.shape
```

```python
# create a random tensor with similar shape to an image tensor

# Create a random tensor of size (224, 224, 3)
random_image_size_tensor = torch.rand(size=(224, 224, 3))
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

```python
# Create a tensor of all zeros
zeros = torch.zeros(size=(3, 4))
zeros, zeros.dtype
```

```python
# Create a tensor of all ones
ones = torch.ones(size=(3, 4))
ones, ones.dtype
```

```python
# Use torch.arange(), torch.range() is deprecated 
zero_to_ten_deprecated = torch.arange(0, 10) # Note: this may return an error in the future

# Create a range of values 0 to 10
zero_to_ten = torch.arange(start=0, end=10, step=1)
zero_to_ten
```

```python
# Can also create a tensor of zeros similar to another tensor
ten_zeros = torch.zeros_like(input=zero_to_ten) # will have same shape
ten_zeros
```

# Tensor Data Types

**Note**: Tensor datatypes is one of the 3 big errors you'll run into with pytorch & deep learning:
1. Tensors not right datatypes.
2. Tensors not right shape
3. Tensors not on the right device 

```python
# Default datatype for tensors is float32
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=None, # defaults to None, which is torch.float32 or whatever datatype is passed
                               device="cuda", # defaults to None, which uses the default tensor type
                               requires_grad=False) # if True, operations performed on the tensor are recorded 

float_32_tensor.shape, float_32_tensor.dtype, float_32_tensor.device
```

```python
float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor.shape, float_16_tensor.dtype
```
